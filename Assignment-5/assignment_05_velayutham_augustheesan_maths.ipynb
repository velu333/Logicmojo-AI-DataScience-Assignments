### Problem-7

A batch of 500 microchips is produced. The probability that a single microchip is defective is 0.02.
Let X be the number of defective microchips in a random sample of 20 microchips from this batch. Assume the sample is taken with replacement, or the batch is large enough that we can model this as independent trials.

(a) What is the probability that exactly 3 microchips in the sample are defective?
(b) What is the probability that at least 2 microchips in the sample are defective?

````python
"""Module to calculate probabilities related to defective microchips using the Binomial distribution."""

from scipy.stats import binom

# Define the probability function and the parameters
# Binomial distribution (Probability of getting exactly k successes in n independent trial):
# P(X=k)  = nCk * p^k * q^(n-k)
# where:
# n = number of independent trials
# p = probability of success (here, a defective chip)
# q = probability of failure
# k = number of successes
#
#
# The Binomial Probability Mass Function (PMF) gives the probability of
# obtaining exactly k successes in n independent Bernoulli trials.

# Parameters
n = 20 # sample size
p = 0.02 # probability that a microchip is defective

# (a) Probability that exactly 3 microchips are defective
k = 3
prob_exactly_3 = binom.pmf(k, n, p)

# (b) Probability that at least 2 microchips are defective
# P(X >= 2) = 1-[P(X=0) - P(X=1)]
prob_at_least_2 = 1 - binom.pmf(0, n, p) - binom.pmf(1, n, p)

#Results
print(f"Probability that exactly 3 chips are defective: {prob_exactly_3:.6f}")
print(f"Probability that at least 2 chips are defective: {prob_at_least_2:.6f}")
````

### Problem-12
The probability density function (pdf) of the triangular distribution is characterized by three points as shown in the figure below.

<img src="https://upload.wikimedia.org/wikipedia/commons/4/45/Triangular_distribution_PMF.png" alt="triangular distribution" width="500"/>

(a) min value of the distribution,
(b) max value of the distribution, and
(c) mode of the distribution

Note: $a \leq c \leq b$

Create a class `TriangularDistribution` with attributes `min`, `max` and `mode`. You need to implement the following methods:

- `pdf(x)`: returns the value of the pdf at point $x$.
- `plot_pdf()`: plots the pdf in the range $[min , max]$
- `cdf(x)`: returns the value of the cdf at point $x$ ; which denotes $P(X \leq x)$
- `plot_cdf()`: plots the cdf in the range $[min , max]$
- `get_probability(x1, x2)`: returns the probability $P(x_1 \leq X \leq x_2)$, where $x_1 \lt x_2$.

Gracefully handle the possible errors and also the edge conditions when either $ a = c $ or $b = c$ in the distribution.

Test your implementation.

````python
"""Module to:
1. Generate a triangular distribution
2. Compute PDF(x) for a given value of x
3. Compute CDF(x) for P(X <= x)
4. Plot the PDF and CDF functions in the range of [min, max]
5. Calculate the probabality P(x1 <= X <= x2)
"""
import matplotlib.pyplot as plt
import numpy as np


class TriangularDistribution:
    """Represents a Triangular probability distribution with parameters (a, b, c)."""

    def __init__(self, a: float, b: float, c: float):
        if not (a <= c <= b):
            raise ValueError("Invalid parameters: Require a <= c <= b.")
        if a == b:
            raise ValueError("Invalid parameters: min and max cannot be the same.")
        self.min = a
        self.max = b
        self.mode = c

    def pdf(self, x: float) -> float:
        """
        Probability Density Function (PDF) of the Triangular distribution.

        f(x) = 0,                        x < a or x > b
              2*(x - a)/((b - a)*(c - a)),  a <= x <= c
              2*(b - x)/((b - a)*(b - c)),  c < x <= b

        Degenerate cases (mode at boundary):
        - Left-degenerate (a == c): f(x) = 2*(b - x)/(b - a)^2
        - Right-degenerate (b == c): f(x) = 2*(x - a)/(b - a)^2
        """
        a, b, c = self.min, self.max, self.mode
        if x < a or x > b:
            return 0.0
        if a == c:  # left-degenerate
            if a <= x <= b:
                return 2 * (b - x) / ((b - a) * (b - c))
        elif b == c:  # right-degenerate
            if a <= x <= b:
                return 2 * (x - a) / ((b - a) * (c - a))
        else:
            if a <= x <= c:
                return 2 * (x - a) / ((b - a) * (c - a))
            elif c < x <= b:
                return 2 * (b - x) / ((b - a) * (b - c))

    def plot_pdf(self):
        """Plot the Probability Density Function (PDF)."""
        x = np.linspace(self.min, self.max, 1000)
        y = [self.pdf(val) for val in x]
        plt.plot(x, y, label="PDF", color='blue')
        plt.title("Triangular Distribution - PDF")
        plt.xlabel("x")
        plt.ylabel("pdf(x)")
        plt.grid(True)
        plt.legend()
        plt.show()

    def cdf(self, x: float) -> float:
        """
        Cumulative Distribution Function (CDF) of the Triangular distribution.

        F(x) = 0,                          x <= a
              (x - a)^2 / ((b - a)*(c - a)),  a < x <= c
              1 - (b - x)^2 / ((b - a)*(b - c)),  c < x <= b
              1,                          x > b

        Degenerate cases (mode at boundary):
        - Left-degenerate (a == c): F(x) = 1 - ((b - x)^2)/(b - a)^2
        - Right-degenerate (b == c): F(x) = ((x - a)^2)/(b - a)^2
        """
        a, b, c = self.min, self.max, self.mode
        if x <= a:
            return 0.0
        elif x >= b:
            return 1.0

        # Degenerate cases
        if a == c:
            return 1 - ((b - x) ** 2) / ((b - a) * (b - c))
        elif b == c:
            return ((x - a) ** 2) / ((b - a) * (c - a))

        # Standard traingular case
        if x <= c:
            return ((x - a) ** 2) / ((b - a) * (c - a))
        else:
            return 1 - ((b - x) ** 2) / ((b - a) * (b - c))

    def plot_cdf(self):
        """Plot the Cumulative Distribution Function (CDF)."""
        x = np.linspace(self.min, self.max, 1000)
        y = [self.cdf(val) for val in x]
        plt.plot(x, y, label="CDF", color='green')
        plt.title("Triangular Distribution - CDF")
        plt.xlabel("x")
        plt.ylabel("cdf(x)")
        plt.grid(True)
        plt.legend()
        plt.show()

    def get_probability(self, x1: float, x2: float) -> float:
        """Compute probability P(x1 <= X <= x2) = F(x2) - F(x1)."""
        if x1 > x2:
            raise ValueError("x1 must be less than x2.")
        return self.cdf(x2) - self.cdf(x1)


# Example usage
if __name__ == "__main__":
    dist = TriangularDistribution(a=0, b=10, c=5)
    print("PDF at x = 3:", dist.pdf(3))
    print("CDF at x = 3:", dist.cdf(3))
    print("P(3 <= X <= 7):", round(dist.get_probability(3, 7), 2))
    dist.plot_pdf()
    dist.plot_cdf()
````

```python
def rand5():
    import numpy as np

```

Write a function `rand7()` that will generate a uniform random integer within the range [1, 7] by only using `rand5()`.

Note: You can call `rand5()` as many time as you want inside the function `rand7()`, but you can not use `np.random.randint(1, 8)` or any other built-in methods in python.

````python
"""Module to generate a uniform random integer within the range [1, 7]
   by only using rand5(), which generates a uniform random integer in [1, 5].
"""
import numpy as np

def rand5():
    """ generate a uniform random integer within the range [1, 5]"""
    return np.random.randint(1, 6)

def rand7():
    """generate a uniform random integer within the range [1, 7]"""
    while True:
        num = (rand5() - 1) * 5 + rand5()  # Generates numbers from 1 to 25
        if num <= 21:
            return (num - 1) % 7 + 1       # Map to [1, 7]

if __name__ == "__main__":
    x = rand7()
    print(f"The generated random integer is: {x}")
````

### Problem-14

A grayscale image can be represented as a matrix where the dimensions are (height, width).
Let's consider a simplified scenario with two grayscale images, $A$ and $B$, both represented as 10×10 matrices where each element represents pixel intensity. Pixel intensity is an integer in the range [0-255].

You want to create a blended image $C$ by combining $A$ and $B$ such that $C=w_1 A+w_2 B$. This represents a weighted average blend.
Note that each element in the resultant matrix $C$ should also fall within the range [0, 255].

write the code of the function `blend_images(image_A, image_B, weight_A, weight_B)` that takes two matrices and two scalar weights and produces another matrix as an output by performing the above operation.

````python
"""
Module to create a blended image from two grayscale images of size 10×10.

Concept:
--------
Blended_Image = weight_A * Image_A + weight_B * Image_B

This is a simple weighted average blend, often used in image processing
to merge two images smoothly.

Requirements:
-------------
- Both input images must have the same shape.
- Pixel intensity range: [0, 255].
"""
import numpy as np
import matplotlib.pyplot as plt


def blend_images(
    image_A:  np.ndarray,
    image_B:  np.ndarray,
    weight_A: float,
    weight_B: float
) -> np.ndarray:


    # Convert to float for weighted operation
    A = np.array(image_A, dtype=np.float32)
    B = np.array(image_B, dtype=np.float32)

    if A.ndim != 2 or B.ndim != 2:
        raise ValueError("Both images must be 2D grayscale arrays.")

    if A.shape != B.shape:
        raise ValueError("Both images must have the same shape.")

    # Normalize the weights
    total_weight = weight_A + weight_B
    weight_A /= total_weight
    weight_B /= total_weight

    # Compute the blended image
    C = weight_A * A + weight_B * B

    # Clip to valid pixel intensity range [0, 255] and convert back to uint8
    C_clipped = np.clip(C, 0, 255).astype(np.uint8)

    return C_clipped


if __name__== "__main__":

    # Create two sample 10x10 grayscale images with random pixel intensities
    image_A = np.random.randint(0, 256, size=(10, 10), dtype=np.uint8)
    image_B = np.random.randint(0, 256, size=(10, 10), dtype=np.uint8)

    # Blend with weights
    blended = blend_images(image_A, image_B, weight_A=0.6, weight_B=0.4)

    # Display the results
    print("Blended Image:\n", blended)

    # Optional Visualization
    plt.subplot(1, 3, 1)
    plt.imshow(image_A, cmap='gray')
    plt.title("Image A")

    plt.subplot(1, 3, 2)
    plt.imshow(image_B, cmap='gray')
    plt.title("Image B")

    plt.subplot(1, 3, 3)
    plt.imshow(blended, cmap='gray')
    plt.title("Blended (0.6A + 0.4B)")

    plt.show()


````

### Problem-15

Calculate the eigen values and eigen vectors of the following matrices.

$A = \begin{bmatrix}
5 & 2 \\
2 & 2
\end{bmatrix}$      

$B = \begin{bmatrix}
1 & 1 & 0\\
0 & 2 & 0\\
-1 & 1 & 3
\end{bmatrix}$

````python
"""
Module to calculate the eigenvalues and eigenvectors of two matrices using NumPy.

- Matrix A: 2x2
- Matrix B: 3x3
- Uses np.linalg.eig
"""

import numpy as np

# Matrix A
A = np.array([[5, 2],
              [2, 2]])

# Calculate the Eigenvalue and Eigenvector of matrix A
eigenvalues_A, eigenvectors_A = np.linalg.eig(A)

# Matrix B
B = np.array([[1, 0, -1],
              [2, 1,  0],
              [0, 3,  3]])

# Calculate the Eigenvalue and Eigenvector of matrix B
eigenvalues_B, eigenvectors_B = np.linalg.eig(B)

# Display the results
print("Eigenvalues of A:", eigenvalues_A)
for i, val in enumerate(eigenvalues_A):
    print(f"Eigenvector corresponding to eigenvalue {val}:\n", eigenvectors_A[:, i])

print("\nEigenvalues of B:", eigenvalues_B)
for i, val in enumerate(eigenvalues_B):
    print(f"Eigenvector corresponding to eigenvalue {val}:\n", eigenvectors_B[:, i])

````
